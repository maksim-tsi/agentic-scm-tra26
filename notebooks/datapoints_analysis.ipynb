# Datapoints Structure Analysis

This notebook analyzes the structure and integrity of the logistics datapoints in `knowledge-base/logistics_datapoints.json` using pandas.

```python
# Import required libraries
import pandas as pd
import json
import os
import openpyxl
```

```python
# Load JSON data into pandas DataFrame
json_path = os.path.join('..', 'knowledge-base', 'logistics_datapoints.json')
with open(json_path, 'r', encoding='utf-8') as f:
    datapoints = json.load(f)
df = pd.DataFrame(datapoints)
df.head()
```

```python
# Inspect DataFrame structure and types
df.info()
```

```python
# Validate required fields
required_fields = [
    'datapoint_id', 'datapoint_type', 'port_area', 'relevant_entity',
    'regulation_category', 'regulation_subcategory', 'regulation_detail', 'source_document'
]
missing_fields = [field for field in required_fields if field not in df.columns]
if missing_fields:
    print(f"Missing required fields: {missing_fields}")
else:
    print("All required fields are present.")
```

```python
# Check for missing or invalid values in key columns
key_columns = required_fields
missing_counts = df[key_columns].isnull().sum()
print("Missing values per key column:")
print(missing_counts)

# Optionally, show rows with missing key values
df_missing = df[df[key_columns].isnull().any(axis=1)]
df_missing.head()
```

```python
# Export DataFrame to Excel
excel_path = os.path.join('..', 'knowledge-base', 'logistics_datapoints.xlsx')
df.to_excel(excel_path, index=False)
print(f"Exported DataFrame to Excel: {excel_path}")
```
